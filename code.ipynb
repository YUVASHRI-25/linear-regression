```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

Fit a simple LR: Study Hours → Exam Score

```python
df=pd.read_csv("study_hours_scores_150.csv")
```

```python
model=LinearRegression()
model.fit(df[["hours"]],df["score"])
y_pred=model.predict(df[["hours"]])
plt.scatter(df["hours"],df["score"])
plt.plot(df["hours"],y_pred,color="red")
plt.show()
```

```python
print(model.coef_[0])
print(model.intercept_)
print(model.score(df[["hours"]],df["score"]))
```

```python
pred=model.predict([[9]])
print(pred[0])
```

Fit LR: House Size → Price

```python
df1=pd.read_csv("house_price_size_150.csv")
model_2=LinearRegression()
model_2.fit(df1[["size_m2"]],df1["price"])
y2_pred=model_2.predict(df1[["size_m2"]])
plt.scatter(df1["size_m2"],df1["price"])
plt.plot(df1["size_m2"],y2_pred,color="red")
plt.show()
```

```python
pred_2=model_2.predict([[120]])
print(pred_2[0])
```

Salary vs Experience

```python
df2=pd.read_csv("salary_experience_150.csv")
model_3=LinearRegression()
model_3.fit(df2[["experience_years"]],df2["salary_k"])
y3_pred=model_3.predict(df2[["experience_years"]])
plt.scatter(df2["experience_years"],df2["salary_k"])
plt.plot(df2["experience_years"],y3_pred,color="red")
plt.show()
```

```python
pred_3=model_3.predict([[7]])[0]
print(pred_3)
```

Multiple Linear Regression: House Price ~ Size + Bedrooms

```python
df3=pd.read_csv("house_price_multifeature_150.csv")
model_4=LinearRegression()
model_4.fit(df3[["size_m2","bedrooms"]],df3["price_k"])
print(model_4.coef_[0])
print(model_4.intercept_)
print(model_4.score(df3[["size_m2","bedrooms"]],df3["price_k"]))
pred_4=model_4.predict([[85,3]])[0]
print(pred_4)
```

Multiple Features: Advertising Spend → Sales

```python
df4=pd.read_csv("advertising_linear_150.csv")
model_5=LinearRegression()
model_5.fit(df4[["TV_spend","radio_spend","social_spend"]],df4["sales_k"])
print(model_5.coef_[0])
print(model_5.intercept_)
print(model_5.score(df4[["TV_spend","radio_spend","social_spend"]],df4["sales_k"]))
pred_5=model_5.predict([[160,32,18]])[0]
print(pred_5)
```
